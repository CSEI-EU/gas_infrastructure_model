```python
df_costs_shares_ESR_2026_run_2021 = dfs['costs_shares_ESR_2026_run_2021']
df_costs_shares_ESR_2026_run_2021 = dfs['costs_shares_ESR_2026_run_2021']
df_costs_shares_ESR_2026_run_2024 = dfs['costs_shares_ESR_2026_run_2024']
df_costs_shares_ESR_2026_run_2024_inv = dfs['costs_shares_ESR_2026_run_2024_inv']
df_costs_shares_ESR_2026_run_2024_plus_no_QA = dfs['costs_shares_ESR_2026_run_2024_plus_no_QA']
df_costs_shares_ESR_2026_run_2024_plus_NO_reduced = dfs['costs_shares_ESR_2026_run_2024_plus_NO_reduced']
df_costs_shares_ESR_2026_run_2024_plus_no_USA = dfs['costs_shares_ESR_2026_run_2024_plus_no_USA']
df_costs_shares_ESR_2026_run_2024_with_RU = dfs['costs_shares_ESR_2026_run_2024_with_RU']
df_costs_shares_ESR_2026_run_2035_AP = dfs['costs_shares_ESR_2026_run_2035_AP']
df_costs_shares_ESR_2026_run_2035_SP = dfs['costs_shares_ESR_2026_run_2035_SP']
```

```python
# Define your reference DataFrame
reference_df = df_costs_shares_ESR_2026_run_2035_SP

# Prepare a dictionary of the others
comparison_dfs = {
    'costs_shares_ESR_2026_run_2024': df_costs_shares_ESR_2026_run_2024,
    #'costs_shares_ESR_2026_run_2021': df_costs_shares_ESR_2026_run_2021,
    #'costs_shares_ESR_2026_run_2024_inv': df_costs_shares_ESR_2026_run_2024_inv,
    #'costs_shares_ESR_2026_run_2024_plus_no_QA': df_costs_shares_ESR_2026_run_2024_plus_no_QA,
    #'costs_shares_ESR_2026_run_2024_plus_NO_reduced': df_costs_shares_ESR_2026_run_2024_plus_NO_reduced,
    #'costs_shares_ESR_2026_run_2024_plus_no_USA': df_costs_shares_ESR_2026_run_2024_plus_no_USA,
    #'costs_shares_ESR_2026_run_2024_with_RU': df_costs_shares_ESR_2026_run_2024_with_RU,
    #'costs_shares_ESR_2026_run_2035_AP': df_costs_shares_ESR_2026_run_2035_AP,
    'costs_shares_ESR_2026_run_2035_SP': df_costs_shares_ESR_2026_run_2035_SP
}

# Compute the differences
difference_dfs = calculate_differences(reference_df, comparison_dfs)
```

```python
summary_total_cost_diff = summarize_total_cost_differences(difference_dfs)
```
