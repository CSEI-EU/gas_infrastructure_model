```python
import pandas as pd
import ast
from collections import defaultdict, deque
import networkx as nx
import os
from cost_and_gas_source_share_functions import *
```

## set the scenario name for the calculation and specify data im/export

```python
#set szenario
szenario_year = '2035'
szenario_name = 'run_' + szenario_year + '_AP'
```

```python
#output specifications
output_file_path = os.path.join('..', '..', '01_data', '02_output_data', '02_unidirectional_results', '01_paper_IAEE', '02_prepared_results')
outout_file_name = '\costs_shares_IAEE_2025_' + szenario_name + '.xlsx'
#create full ouput paths
output_file_path_excel  = output_file_path + outout_file_name
full_output_file_path = os.path.abspath(os.path.join(os.getcwd(), output_file_path_excel))
```

```python
# Specify the path to the results Excel file
input_results_file_path_results = os.path.join('..', '..','01_data', '02_output_data', '02_unidirectional_results', '01_paper_IAEE', '01_raw_results')
results_file_name = '\outputs_IAEE_2025_' + szenario_name + '.xlsx'


# Specify the path to the input data Excel file
input_file_path = os.path.join('..', '..','01_data', '01_input_data', '02_processed', '01_paper_IAEE')
input_file_name = '\inputs_IAEE_2025_' + szenario_name + '.xlsx'

# Specify the path to the raw edges data Excel file
file_name_edges_cap_cost_raw = '\inputs_IAEE_2025_edges_raw.xlsx'

#create full input paths
#results of the optimization
input_results_file_path_excel  = input_results_file_path_results + results_file_name
full_input_results_path = os.path.abspath(os.path.join(os.getcwd(), input_results_file_path_excel))

#inputs of the optimization
input_file_path_excel  = input_file_path + input_file_name
full_input_file_path = os.path.abspath(os.path.join(os.getcwd(), input_file_path_excel))

#edges raw path
file_path_edges_cap_costs_raw  = input_file_path + file_name_edges_cap_cost_raw
full_file_path_edges_cap_costs_raw = os.path.abspath(os.path.join(os.getcwd(), file_path_edges_cap_costs_raw))
```

## read in data

```python
df_flows_raw = pd.read_excel(full_input_results_path, sheet_name='flows_methane_edges')
df_edges_raw = pd.read_excel(full_input_file_path, sheet_name='Parameters')
df_edges_cap_cost_raw = pd.read_excel(full_file_path_edges_cap_costs_raw, sheet_name='edges_raw')
```

## Prepare the data for the calculations

```python
#set real cost values for transportation (some edges have artifical cost in the optimization)
df_edges_raw_updated = update_edge_costs_to_real_value(df_edges_raw, df_edges_cap_cost_raw, commodity='Methane')
```

```python
#Clean the flows data frame for rows with 0's 
df_flows_raw = df_flows_raw[df_flows_raw['Flow'] != 0].reset_index(drop=True)
```

### Calculate the cost and shares

```python
df_costs, df_source_shares = compute_average_costs(df_flows_raw, df_edges_raw_updated, commodity='Methane')
```

### reduce the results to the relevant information

```python
# Keep only nodes with no underscore (pure region names)
df_costs_cleaned = df_costs[~df_costs['Node'].str.contains('_')].copy()

# Drop the 'Total Flow' column
df_costs_cleaned = df_costs_cleaned.drop(columns=['Total Flow'])

# Reset index
df_costs_cleaned = df_costs_cleaned.reset_index(drop=True)
```

```python
# Keep only rows where 'Node' has no underscore (i.e., pure country nodes)
df_shares_cleaned = df_source_shares[~df_source_shares['Node'].str.contains('_')].copy()

# Drop the 'Flow' column
df_shares_cleaned = df_shares_cleaned.drop(columns=['Flow'])

# Reset index
df_shares_cleaned = df_shares_cleaned.reset_index(drop=True)
```

## Export the results

```python
# Export to Excel with multiple sheets
with pd.ExcelWriter(full_output_file_path, engine='openpyxl') as writer:
    df_costs_cleaned.to_excel(writer, index=False, sheet_name='cost')
    df_shares_cleaned.to_excel(writer, index=False, sheet_name='shares')
```
