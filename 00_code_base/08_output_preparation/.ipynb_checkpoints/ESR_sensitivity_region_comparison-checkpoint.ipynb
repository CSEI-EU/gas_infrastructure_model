# ESR 2024 Sensitivity Comparison by Region

This notebook compares the **2024 base case** against multiple sensitivity runs and calculates the percentage change for selected European country groups.

It supports the following sensitivity types:
- `lng`
- `pipe`
- `prod`
- `regas`
- `liqu`
- `channel`

For each sensitivity type, the notebook looks for:
- `10pup`
- `10pdown`
- `05pup`
- `05pdown`

The country-region definitions are read from `country_regions_ESR_2024.xlsx`.

```python
from pathlib import Path
import pandas as pd

# ============================================================
# 1. FILE LOCATIONS
# ============================================================

# Main 2024 base case
BASE_FILE = Path("/mnt/data/costs_shares_ESR_2026_run_2024.xlsx")

# Excel file containing the country-to-region definitions
REGION_FILE = Path("/mnt/data/country_regions_ESR_2024.xlsx")

# Folder containing the sensitivity workbooks
SENSITIVITY_FOLDER = Path("/mnt/data")

# Excel sheet containing the cost data
COST_SHEET = "cost"

# Columns for which percentage changes should be calculated
VALUE_COLUMNS = [
    "Gas Cost",
    "Transport Cost",
    "Total Cost",
]
```

```python
# ============================================================
# 2. SENSITIVITY DEFINITIONS
# ============================================================

SENSITIVITY_TYPES = [
    "lng",
    "pipe",
    "prod",
    "regas",
    "liqu",
    "channel",
]

SENSITIVITY_LEVELS = [
    "10pup",
    "10pdown",
    "05pup",
    "05pdown",
]
```

```python
# ============================================================
# 3. LOAD COUNTRY GROUPS FROM EXCEL
# ============================================================

def load_country_groups(region_file: Path) -> dict[str, list[str]]:
    """
    Read region definitions from the 'regions' sheet.

    Expected structure:
        Node | Country | Europe | North-Eastern Europe | ...

    A value of 1 means that the node belongs to the corresponding
    region. This allows regions to overlap.
    """

    df = pd.read_excel(region_file, sheet_name="regions")

    required_columns = {"Node", "Country"}
    missing = required_columns - set(df.columns)
    if missing:
        raise ValueError(f"Missing required columns in region file: {missing}")

    region_columns = [c for c in df.columns if c not in ["Node", "Country"]]
    if not region_columns:
        raise ValueError("No region columns found in the region file.")

    df["Node"] = df["Node"].astype(str).str.strip()

    country_groups = {}
    for region in region_columns:
        mask = pd.to_numeric(df[region], errors="coerce").fillna(0).eq(1)
        country_groups[region] = df.loc[mask, "Node"].tolist()

    return country_groups


country_groups = load_country_groups(REGION_FILE)

print("Country groups loaded:")
for region, countries in country_groups.items():
    print(f"{region}: {', '.join(countries)}")
```

```python
# ============================================================
# 4. FIND SENSITIVITY FILES
# ============================================================

def find_sensitivity_files(
    folder: Path,
    sensitivity_types: list[str],
    sensitivity_levels: list[str],
) -> dict[str, Path]:
    """
    Find sensitivity workbooks using the filename convention:

        *sens_<type>_cost_<level>*.xlsx

    Example:
        costs_shares_ESR_2026_run_2024_sens_regas_cost_05pdown.xlsx
    """

    sensitivity_files = {}

    for sensitivity_type in sensitivity_types:
        for level in sensitivity_levels:

            pattern = f"*sens_{sensitivity_type}_cost_{level}*.xlsx"
            matches = sorted(folder.glob(pattern))

            # Exclude the base workbook if a broad pattern ever catches it
            matches = [m for m in matches if m.resolve() != BASE_FILE.resolve()]

            key = f"{sensitivity_type}_{level}"

            if not matches:
                print(f"WARNING: No file found for {key} (pattern: {pattern})")
                continue

            if len(matches) > 1:
                print(f"WARNING: Multiple files found for {key}; using: {matches[0].name}")
                for match in matches[1:]:
                    print(f"         ignored: {match.name}")

            sensitivity_files[key] = matches[0]

    return sensitivity_files


sensitivity_files = find_sensitivity_files(
    SENSITIVITY_FOLDER,
    SENSITIVITY_TYPES,
    SENSITIVITY_LEVELS,
)

print("\nSensitivity files found:")
for name, file in sensitivity_files.items():
    print(f"{name:25s} -> {file.name}")
```

```python
# ============================================================
# 5. READ COST TABLES
# ============================================================

def read_cost_table(file: Path) -> pd.DataFrame:
    """Read and lightly standardise the cost table."""

    df = pd.read_excel(file, sheet_name=COST_SHEET)

    if "Node" not in df.columns:
        raise ValueError(f"No 'Node' column found in {file.name}")

    missing_value_columns = [c for c in VALUE_COLUMNS if c not in df.columns]
    if missing_value_columns:
        raise ValueError(
            f"Missing columns {missing_value_columns} in {file.name}"
        )

    df["Node"] = df["Node"].astype(str).str.strip()

    for col in VALUE_COLUMNS:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return df


base_df = read_cost_table(BASE_FILE)

sensitivity_dfs = {
    name: read_cost_table(file)
    for name, file in sensitivity_files.items()
}

print(f"Base rows: {len(base_df)}")
print(f"Sensitivity cases loaded: {len(sensitivity_dfs)}")
```

```python
# ============================================================
# 6. REGION AGGREGATION AND PERCENTAGE CHANGE
# ============================================================

def aggregate_region(
    df: pd.DataFrame,
    countries: list[str],
    value_column: str,
) -> float:
    """
    Sum the selected cost variable over all nodes in a region.
    """

    subset = df[df["Node"].isin(countries)]
    return subset[value_column].sum(min_count=1)


def percentage_change(base_value: float, sensitivity_value: float) -> float:
    """Calculate percentage change relative to the base case."""

    if pd.isna(base_value) or pd.isna(sensitivity_value) or base_value == 0:
        return float("nan")

    return (sensitivity_value / base_value - 1.0) * 100.0


def compare_sensitivities(
    base_df: pd.DataFrame,
    sensitivity_dfs: dict[str, pd.DataFrame],
    country_groups: dict[str, list[str]],
    value_columns: list[str],
) -> dict[str, pd.DataFrame]:
    """
    Compare the base case against every sensitivity case for
    every region and requested value column.

    The returned DataFrame contains the base value plus the
    percentage change for each sensitivity.
    """

    results = {}

    for value_column in value_columns:
        rows = []

        for region, countries in country_groups.items():

            base_value = aggregate_region(
                base_df, countries, value_column
            )

            row = {
                "Region": region,
                "Base": base_value,
            }

            for sensitivity_name, sensitivity_df in sensitivity_dfs.items():
                sensitivity_value = aggregate_region(
                    sensitivity_df, countries, value_column
                )

                row[sensitivity_name] = percentage_change(
                    base_value, sensitivity_value
                )

            rows.append(row)

        results[value_column] = pd.DataFrame(rows)

    return results


results = compare_sensitivities(
    base_df=base_df,
    sensitivity_dfs=sensitivity_dfs,
    country_groups=country_groups,
    value_columns=VALUE_COLUMNS,
)
```

```python
# ============================================================
# 7. DISPLAY RESULTS
# ============================================================

for cost_type, df in results.items():
    print("\n" + "=" * 110)
    print(cost_type)
    print("=" * 110)

    display(
        df.style
        .format({"Base": "{:,.2f}"}, na_rep="-")
        .format(
            {
                c: "{:+.2f}%"
                for c in df.columns
                if c != "Region" and c != "Base"
            },
            na_rep="-",
        )
    )
```

## Optional: create one combined output table

The following cell combines the results for all cost variables into a single Excel workbook. Each cost variable is written to a separate sheet.


```python
# ============================================================
# 8. OPTIONAL: EXPORT RESULTS TO EXCEL
# ============================================================

OUTPUT_FILE = Path(
    "/mnt/data/ESR_2024_regional_sensitivity_comparison.xlsx"
)

with pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl") as writer:
    for cost_type, df in results.items():
        # Excel sheet names cannot exceed 31 characters
        sheet_name = cost_type[:31]
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"Results written to: {OUTPUT_FILE}")
```
