# Script to get the differences of emissions between scenarios

This jupyter notebook contains all routine calculations of the emission differences in Europe compared to another scenario. 

**Authors:** Johannes Giehl (jfg.eco@cbs.dk)

## Import packages

```python
#import
import pandas as pd
from cost_and_gas_source_share_functions import *
from emission_calculation_functions import *
```

## Define path to data

```python
#path for output and input data
#for gas shares per country
data_file_path = os.path.join('..', '..', '01_data', '02_output_data', '02_unidirectional_results', '02_paper_ESR', '02_prepared_results')
full_data_file_path = os.path.abspath(os.path.join(os.getcwd(), data_file_path))
#for demand per country
demand_data_path = os.path.join('..', '..', '01_data', '01_input_data', '02_processed', '02_paper_ESR')
full_demand_data_path = os.path.abspath(os.path.join(os.getcwd(), demand_data_path))
#emission factors
emission_factors_data_path = os.path.join('..', '..', '01_data', '01_input_data', '02_processed')
emission_factors_file_name = '\emission_factors_natural_gas.xlsx'
emission_factors_path = emission_factors_data_path + emission_factors_file_name
full_emission_factors_path = os.path.abspath(os.path.join(os.getcwd(), emission_factors_path))

#specify the output file
output_file_name_totals = '\emission_differences.xlsx'
output_file_name_ef = '\emission_factors.xlsx'
output_file_path_totals = data_file_path + output_file_name_totals
output_file_path_ef = data_file_path + output_file_name_ef
full_output_file_path_totals = os.path.abspath(os.path.join(os.getcwd(), output_file_path_totals))
full_output_file_path_ef = os.path.abspath(os.path.join(os.getcwd(), output_file_path_ef))
```

## Load te data

```python
#get all cost dfs from the excels in the prepared results folder
#shares of gas sources per node
dfs = load_excel_sheets_by_name(full_data_file_path, "costs_shares", sheet_name="shares")
#demand of the nodes
dfs_demand = load_excel_sheets_by_name(full_demand_data_path, "inputs_ESR_2026_run", sheet_name="Supply")
#emission factors imports
dfs_emission_factors_im = pd.read_excel(full_emission_factors_path, sheet_name="Emission_Factors_Import", index_col=0)
#emission factors domestic
dfs_emission_factors_dom = pd.read_excel(full_emission_factors_path, sheet_name="Emission_Factors_Domestic", index_col=0)
```

```python
#get df with information of EU membership
dfs_EU_countries = dfs_emission_factors_im.iloc[:, 0:1]
```

## Scenario names

```python
df_names_demand = list(dfs_demand.keys())
df_names_demand
```

```python
df_names = list(dfs.keys())
df_names
```

```python
#create list of scenario names
scenarios = extract_scenario_names(df_names_demand, df_names)
```

## Prepare individual dataframes

```python
#dataframe for the shares
for scenario in scenarios:
    var_name = f"df_shares_{scenario}"
    key = f"costs_shares_{scenario}"
    
    globals()[var_name] = dfs[key]
```

```python
#dataframe for the demand
for scenario in scenarios:
    var_name = f"df_demand_{scenario}"
    key = f"inputs_{scenario}"
    
    if key not in dfs_demand:
        print(f"Warning: {key} not found in dfs_demand")
        continue
    
    globals()[var_name] = dfs_demand[key]
```

```python
#filter the dataframes so that they contain only demand
df_demand_2019_filtered = filter_negative_supply(df_demand_ESR_2026_run_2019, "Methane")
df_demand_2021_filtered = filter_negative_supply(df_demand_ESR_2026_run_2021, "Methane")
df_demand_2024_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024, "Methane")
df_demand_2024_inv_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024_inv, "Methane")
df_demand_2024_plus_no_QA_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024_plus_no_QA, "Methane")
df_demand_2024_plus_NO_reduced_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024_plus_NO_reduced, "Methane")
df_demand_2024_plus_no_USA_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024_plus_no_USA, "Methane")
df_demand_2024_with_RU_filtered = filter_negative_supply(df_demand_ESR_2026_run_2024_with_RU, "Methane")
df_demand_2035_AP_filtered = filter_negative_supply(df_demand_ESR_2026_run_2035_AP, "Methane")
df_demand_2035_SP_filtered = filter_negative_supply(df_demand_ESR_2026_run_2035_SP, "Methane")
```

## Calculate emissions

```python
#import emissions
df_import_emissions_2019 = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2019, df_demand_2019_filtered)
df_import_emissions_2021 = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2021, df_demand_2021_filtered)
df_import_emissions_2024 = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024, df_demand_2024_filtered)
df_import_emissions_2024_inv = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024_inv, df_demand_2024_inv_filtered)
df_import_emissions_2024_no_QA = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024_plus_no_QA, df_demand_2024_plus_no_QA_filtered)
df_import_emissions_2024_NO_red = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024_plus_NO_reduced, df_demand_2024_plus_NO_reduced_filtered)
df_import_emissions_2024_no_USA = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024_plus_no_USA, df_demand_2024_plus_no_USA_filtered)
df_import_emissions_2024_with_RU = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2024_with_RU, df_demand_2024_with_RU_filtered)
df_import_emissions_2035_AP = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2035_AP, df_demand_2035_AP_filtered)
df_import_emissions_2035_SP = calculate_emissions_per_country(dfs_emission_factors_im, df_shares_ESR_2026_run_2035_SP, df_demand_2035_SP_filtered)
```

```python
#domestic emissions
df_domestic_emissions_2019 = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2019, df_demand_2019_filtered)
df_domestic_emissions_2021 = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2021, df_demand_2021_filtered)
df_domestic_emissions_2024 = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024, df_demand_2024_filtered)
df_domestic_emissions_2024_inv = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024_inv, df_demand_2024_inv_filtered)
df_domestic_emissions_2024_no_QA = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024_plus_no_QA, df_demand_2024_plus_no_QA_filtered)
df_domestic_emissions_2024_NO_red = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024_plus_NO_reduced, df_demand_2024_plus_NO_reduced_filtered)
df_domestic_emissions_2024_no_USA = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024_plus_no_USA, df_demand_2024_plus_no_USA_filtered)
df_domestic_emissions_2024_with_RU = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2024_with_RU, df_demand_2024_with_RU_filtered)
df_domestic_emissions_2035_AP = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2035_AP, df_demand_2035_AP_filtered)
df_domestic_emissions_2035_SP = calculate_domestic_emissions(dfs_emission_factors_dom, df_shares_ESR_2026_run_2035_SP, df_demand_2035_SP_filtered)
```

```python
#emissions for al Europen countries
df_total_emissions_2019 = combine_total_emissions(df_import_emissions_2019, df_domestic_emissions_2019)
df_total_emissions_2021 = combine_total_emissions(df_import_emissions_2021, df_domestic_emissions_2021)
df_total_emissions_2024 = combine_total_emissions(df_import_emissions_2024, df_domestic_emissions_2024)
df_total_emissions_2024_inv = combine_total_emissions(df_import_emissions_2024_inv, df_domestic_emissions_2024_inv)
df_total_emissions_2024_plus_no_QA  = combine_total_emissions(df_import_emissions_2024_no_QA, df_domestic_emissions_2024_no_QA)
df_total_emissions_2024_plus_NO_reduced = combine_total_emissions(df_import_emissions_2024_NO_red, df_domestic_emissions_2024_NO_red)
df_total_emissions_2024_plus_no_USA = combine_total_emissions(df_import_emissions_2024_no_USA, df_domestic_emissions_2024_no_USA)
df_total_emissions_2024_with_RU = combine_total_emissions(df_import_emissions_2024_with_RU, df_domestic_emissions_2024_with_RU)
df_total_emissions_2035_AP = combine_total_emissions(df_import_emissions_2035_AP, df_domestic_emissions_2035_AP)
df_total_emissions_2035_SP = combine_total_emissions(df_import_emissions_2035_SP, df_domestic_emissions_2035_SP)
```

## Emission Factors of the scenarios

```python
#Europe
avg_ef_results = []

for scenario in scenarios:
    suffix = get_scenario_suffix(scenario)

    total_var = f"df_total_emissions_{suffix}"
    demand_var = f"df_demand_{suffix}_filtered"

    if total_var not in globals():
        print(f"Warning: {total_var} not found")
        continue

    if demand_var not in globals():
        print(f"Warning: {demand_var} not found")
        continue

    avg_ef_results.append(
        calculate_average_emission_factor_from_totals(
            scenario,
            globals()[total_var],
            globals()[demand_var]
        )
    )
```

```python
df_avg_ef_scenarios = pd.DataFrame(
    avg_ef_results,
    columns=["Scenario", "tCO2e/GWh in Europe"]
).set_index("Scenario")
df_avg_ef_scenarios
```

```python
#EU
avg_ef_results_eu = []

for scenario in scenarios:
    suffix = get_scenario_suffix(scenario)

    total_var = f"df_total_emissions_{suffix}"
    demand_var = f"df_demand_{suffix}_filtered"

    if total_var not in globals():
        print(f"Warning: {total_var} not found")
        continue

    if demand_var not in globals():
        print(f"Warning: {demand_var} not found")
        continue

    avg_ef_results_eu.append(
        calculate_average_emission_factor_from_totals(
            scenario,
            globals()[total_var],
            globals()[demand_var],
            only_EU=True,
            dfs_EU_countries=dfs_EU_countries
        )
    )
```

```python
df_avg_ef_scenarios_eu = pd.DataFrame(
    avg_ef_results_eu,
    columns=["Scenario", "tCO2e/GWh in the EU"]
).set_index("Scenario")
df_avg_ef_scenarios_eu
```
